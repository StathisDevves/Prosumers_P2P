```python
#!/usr/bin/env python3
"""Build the EQ1–EQ32 Daily 24-hour prosumer/P2P Excel model.

This generator uses openpyxl and writes native Excel formulas into the workbook.
The optimization horizon is fixed by calendar day: rows 2–25, 26–49, ..., 8738–8761.
It is not a rolling/sliding 24-hour window.

Expected inputs in the same directory as this script:
    Input Data(2).xlsx
    Input Core assumptions(2).xlsx   (optional; retained as a documented source)

Output:
    Output_EQ1_EQ32_DailyOptimization.xlsx

Requirements:
    Python 3.10+
    openpyxl >= 3.1
"""

from __future__ import annotations

from pathlib import Path
from typing import Dict, Iterable, Tuple

from openpyxl import load_workbook
from openpyxl.styles import Alignment, Border, Font, PatternFill, Side
from openpyxl.utils import get_column_letter
from openpyxl.worksheet.dimensions import ColumnDimension


BASE_DIR = Path.cwd()
INPUT_DATA_FILE = BASE_DIR / "Input Data.xlsx"
INPUT_ASSUMPTIONS_FILE = BASE_DIR / "Input Core assumptions.xlsx"
OUTPUT_FILE = BASE_DIR / "Output_EQ1_EQ32_DailyOptimization.xlsx"

DATA_SHEET = "Solar_8760_2025"
FIRST_DATA_ROW = 2
HOURS_PER_DAY = 24
DAYS_PER_YEAR = 365
LAST_DATA_ROW = FIRST_DATA_ROW + HOURS_PER_DAY * DAYS_PER_YEAR - 1

# Existing input columns assumed by the source workbook:
# H = hourly PV production (kWh)
# I = baseline demand (kWh)
# J = spot price (EUR/MWh), converted to EUR/kWh in L


PARAMETERS = [
    ("Utility curvature", "alpha", 0.25, "€/kWh²", "EQ1–EQ8, EQ19–EQ30", "User-selected"),
    ("Load-shifting coefficient", "zeta", 0.20, "€/kWh²", "EQ8, EQ14, EQ21, EQ24–EQ29", "User-selected"),
    ("Battery degradation coefficient", "psi", 0.03, "€/kWh", "EQ10, EQ14, EQ21, EQ24–EQ29", "User-selected"),
    ("Battery capacity", "C_bat", 30.0, "kWh", "EQ2, EQ4, EQ9, EQ12", "Core assumption"),
    ("Maximum charging power", "P_ch_max", 3.0, "kWh/h", "EQ2, EQ11", "One-hour time step"),
    ("Maximum discharging power", "P_dis_max", 3.0, "kWh/h", "EQ4, EQ11, EQ28–EQ31", "One-hour time step"),
    ("Charging efficiency", "eta_ch", 0.92, "fraction", "EQ9, EQ12", "One-way efficiency"),
    ("Discharging efficiency", "eta_dis", 0.92, "fraction", "EQ9, EQ12", "One-way efficiency"),
    ("Initial SOC", "SOC_0", 0.90, "fraction of capacity", "EQ9, EQ12", "90% of capacity"),
    ("Minimum SOC", "SOC_min", 0.10, "fraction of capacity", "EQ4, EQ9, EQ12", "Operational bound"),
    ("Maximum SOC", "SOC_max", 1.00, "fraction of capacity", "EQ2, EQ9, EQ12", "Operational bound"),
    ("Local supply-demand ratio", "SDR", 0.50, "fraction", "EQ13", "Default/calibration value"),
    ("Baseline network cost", "tau_base", 0.05, "€/kWh", "EQ17", "Lower stated bound"),
    ("Green preference premium", "theta", 0.022, "€/kWh", "EQ15, EQ19, EQ21, EQ23–EQ27", "Given"),
    ("Green guarantee indicator", "rho", 1.00, "fraction", "EQ15, EQ19, EQ21, EQ23–EQ27", "1 = fully green"),
    ("Prosumer marginal PV cost", "b_i", 0.065, "€/kWh", "EQ24–EQ27", "Range midpoint"),
    ("Prosumer quadratic cost", "a_i", 0.025, "€/kWh²", "EQ24–EQ27", "Range midpoint"),
    ("Network scaling coefficient at minimum load", "a1_min", 0.10, "fraction", "EQ18", "At minimum load"),
    ("Network scaling coefficient at maximum load", "a1_max", 0.50, "fraction", "EQ18", "At maximum load"),
    ("Low-price threshold share", "Low_share", 1 / 3, "fraction", "EQ2 decision rule", "Lower third of each day price range"),
    ("High-price threshold share", "High_share", 2 / 3, "fraction", "EQ4 decision rule", "Upper third of each day price range"),
]

HEADERS = {
    "K": "ω_t (€/kWh)",
    "L": "Price_spot (€/kWh)",
    "M": "Daily Price Min (€/kWh)",
    "N": "Daily Price Max (€/kWh)",
    "O": "Daily Low Threshold (€/kWh)",
    "P": "Daily High Threshold (€/kWh)",
    "Q": "Daily PV Max (kWh)",
    "R": "Decision",
    "S": "E_self,n,t EQ1 (kWh)",
    "T": "E_charge (kWh)",
    "U": "E_discharge (kWh)",
    "V": "SOC_begin (kWh)",
    "W": "SOC_end EQ9/EQ12 (kWh)",
    "X": "E_grid,t EQ2–EQ4 (kWh)",
    "Y": "ε_t Elasticity EQ6 (-)",
    "Z": "U_cons EQ7 (€)",
    "AA": "E_actual EQ22 (kWh)",
    "AB": "ΔE (kWh)",
    "AC": "D_shift (€)",
    "AD": "E_bat EQ11 (kWh)",
    "AE": "C_deg EQ10 (€)",
    "AF": "E_sell (kWh)",
    "AG": "E_buy (kWh)",
    "AH": "W_spot (€)",
    "AI": "Utility EQ8/EQ29 (€)",
    "AJ": "Daily U_max EQ8/EQ29 (€)",
    "AK": "SDR",
    "AL": "Daily Charge Reference Price (€/kWh)",
    "AM": "Price_P2P EQ13 (€/kWh)",
    "AN": "E_inj EQ16 (kWh)",
    "AO": "U_prosumer EQ14 (€)",
    "AP": "a1 EQ18 (-)",
    "AQ": "L_total/C_grid (-)",
    "AR": "γ EQ18 (-)",
    "AS": "τ EQ17 (€/kWh)",
    "AT": "Green benefit (€)",
    "AU": "SW_total EQ15 (€)",
    "AV": "U_buyer EQ19 (€)",
    "AW": "A EQ21 (€)",
    "AX": "Price_eq EQ20 (€/kWh)",
    "AY": "CS EQ23 (€)",
    "AZ": "PS EQ24 (€)",
    "BA": "LC EQ26 (€)",
    "BB": "QC EQ27 (€)",
    "BC": "SW_P2P EQ25 (€)",
    "BD": "CS_pros EQ28 (€)",
    "BE": "CS_t1 EQ30 (€)",
    "BF": "Daily Price Peak (€/kWh)",
    "BG": "GS_grid EQ31 (€)",
    "BH": "SW_grid EQ32 (€)",
}


EQ_GUIDE = [
    ("EQ1", "Self-consumption decision", "E_self,n,t", "S", "kWh", "Constrained quadratic-utility optimum and available PV/demand."),
    ("EQ2–EQ4", "Grid/battery operational rules", "E_grid, E_charge, E_discharge", "T,U,X", "kWh", "Fixed daily low/high price thresholds."),
    ("EQ6", "Price elasticity", "ε_t", "Y", "dimensionless", "Calculated from spot price, alpha and grid demand."),
    ("EQ7", "Consumer utility", "U_cons", "Z", "€", "Quadratic utility of self-consumed energy."),
    ("EQ8", "Daily utility maximization", "U_max", "AI,AJ", "€", "Hourly utility plus fixed 24-hour daily aggregation."),
    ("EQ9/EQ12", "Battery SOC recursion", "SOC_begin, SOC_end", "V,W", "kWh", "SOC bounded by minimum and maximum shares of capacity."),
    ("EQ10", "Battery degradation cost", "C_deg", "AE", "€", "Throughput proxy psi × |E_bat|."),
    ("EQ11", "Battery net flow", "E_bat", "AD", "kWh", "Charge positive, discharge negative."),
    ("EQ13", "P2P price", "Price_P2P", "AM", "€/kWh", "Spot/reference-price and SDR-based expression."),
    ("EQ14", "Prosumer utility", "U_prosumer", "AO", "€", "Utility including P2P injection revenue."),
    ("EQ15", "Total social welfare", "SW_total", "AU", "€", "Consumer utility, green benefit and network charge."),
    ("EQ16", "P2P injection", "E_inj", "AN", "kWh", "PV surplus after self-consumption and charging."),
    ("EQ17", "Network tariff", "τ", "AS", "€/kWh", "Load-dependent network tariff."),
    ("EQ18", "Network scaling", "a1, γ", "AP,AR", "dimensionless", "Daily load-normalized scaling."),
    ("EQ19", "Buyer utility", "U_buyer", "AV", "€", "Utility from P2P energy net of P2P payment."),
    ("EQ20", "Equilibrium price", "Price_eq", "AX", "€/kWh", "Derived from the bilateral welfare expression."),
    ("EQ21", "Auxiliary welfare term", "A", "AW", "€", "Intermediate term used for equilibrium pricing."),
    ("EQ22", "Actual consumption", "E_actual", "AA", "kWh", "Self-consumption + discharge + grid purchase."),
    ("EQ23", "Consumer surplus", "CS", "AY", "€", "Gross utility/green value less P2P expenditure."),
    ("EQ24", "Producer surplus", "PS", "AZ", "€", "P2P revenue less production, degradation and shifting costs."),
    ("EQ25", "P2P social welfare", "SW_P2P", "BC", "€", "Linear contribution less quadratic cost."),
    ("EQ26", "Linear component", "LC", "BA", "€", "Linear component of EQ25."),
    ("EQ27", "Quadratic component", "QC", "BB", "€", "Quadratic component of EQ25."),
    ("EQ28", "Prosumer consumer surplus", "CS_pros", "BD", "€", "Decision-dependent prosumer surplus."),
    ("EQ29", "Daily maximum utility", "U_max", "AI,AJ", "€", "Same fixed calendar-day aggregation as EQ8."),
    ("EQ30", "Period-1 surplus", "CS_t1", "BE", "€", "Discharge-state surplus expression."),
    ("EQ31", "Grid-service gain", "GS_grid", "BG", "€", "Peak-price spread × discharge."),
    ("EQ32", "Grid social welfare", "SW_grid", "BH", "€", "Prosumer surplus plus grid-service gain."),
]


THIN_GREY = Side(style="thin", color="B7C9D6")
HEADER_FILL = PatternFill("solid", fgColor="D9EAF7")
TITLE_FILL = PatternFill("solid", fgColor="1F4E78")
SECTION_FILL = PatternFill("solid", fgColor="EAF2F8")


def day_bounds(row: int) -> Tuple[int, int]:
    """Return the first and last worksheet rows of the fixed calendar day containing row."""
    day_index = (row - FIRST_DATA_ROW) // HOURS_PER_DAY
    start = FIRST_DATA_ROW + day_index * HOURS_PER_DAY
    return start, start + HOURS_PER_DAY - 1


def formula_map(row: int) -> Dict[str, str]:
    """Return all native Excel formulas for one hourly row."""
    start, end = day_bounds(row)
    r = row
    return {
        "K": f'=IF(OR(F{r}<=6,F{r}=24),0.55,IF(F{r}<=10,1,IF(F{r}<=16,0.8,IF(F{r}<=22,1.35,0.65))))',
        "L": f'=J{r}/1000',
        "M": f'=MIN($L${start}:$L${end})',
        "N": f'=MAX($L${start}:$L${end})',
        "O": f'=M{r}+(N{r}-M{r})*Parameters!$C$22',
        "P": f'=M{r}+(N{r}-M{r})*Parameters!$C$23',
        "Q": f'=MAX($H${start}:$H${end})',
        "S": f'=MAX(0,MIN(I{r},H{r},(K{r}-L{r})/Parameters!$C$3))',
        "T": f'=IF(AND(H{r}>S{r},L{r}<=O{r},V{r}<Parameters!$C$6*Parameters!$C$13),MAX(0,MIN(Parameters!$C$7,H{r}-S{r},(Parameters!$C$6*Parameters!$C$13-V{r})/Parameters!$C$9)),0)',
        "U": f'=IF(AND(I{r}>S{r},L{r}>=P{r},V{r}>Parameters!$C$6*Parameters!$C$12),MAX(0,MIN(Parameters!$C$8,I{r}-S{r},(V{r}-Parameters!$C$6*Parameters!$C$12)*Parameters!$C$10)),0)',
        "R": f'=IF(I{r}-S{r}-U{r}>0,"Buy from the Grid",IF(AND(U{r}>0,S{r}+U{r}>=I{r}),"discharge",IF(AND(H{r}>S{r},T{r}>0),"charge",IF(H{r}>S{r},"sell","self consume"))))',
        "V": '=Parameters!$C$6*Parameters!$C$11' if r == FIRST_DATA_ROW else f'=W{r-1}',
        "W": f'=MAX(Parameters!$C$6*Parameters!$C$12,MIN(Parameters!$C$6*Parameters!$C$13,V{r}+Parameters!$C$9*T{r}-U{r}/Parameters!$C$10))',
        "X": f'=MAX(0,I{r}-S{r}-U{r})',
        "Y": f'=IF(X{r}=0,0,-(1/Parameters!$C$3)*L{r}/X{r})',
        "Z": f'=K{r}*S{r}-Parameters!$C$3/2*S{r}^2',
        "AA": f'=S{r}+U{r}+AG{r}',
        "AB": f'=I{r}-AA{r}',
        "AC": f'=Parameters!$C$4/2*AB{r}^2',
        "AD": f'=T{r}-U{r}',
        "AE": f'=Parameters!$C$5*ABS(AD{r})',
        "AF": f'=IF(T{r}>0,MAX(0,H{r}-S{r}-MIN(Parameters!$C$7,MAX(0,Parameters!$C$6*Parameters!$C$13-V{r}),MAX(0,H{r}-S{r}))),MAX(0,H{r}-S{r}))',
        "AG": f'=MAX(0,I{r}-S{r}-U{r})',
        "AH": f'=L{r}*(AF{r}-AG{r})',
        "AI": f'=Z{r}-AC{r}+AH{r}-AE{r}',
        "AJ": f'=SUM($AI${start}:$AI${end})',
        "AK": '=Parameters!$C$14',
        "AL": f'=MIN($L${start}:$L${end})',
        "AM": f'=IF(AF{r}=0,L{r},IFERROR(L{r}*AL{r}/(L{r}-AL{r})*AK{r}+AL{r},(L{r}+AL{r})/2))',
        "AN": f'=AF{r}',
        "AO": f'=Z{r}-AC{r}+AN{r}*AM{r}-AE{r}',
        "AP": f'=Parameters!$C$20+(Parameters!$C$21-Parameters!$C$20)*IF(MAX($I${start}:$I${end})=MIN($I${start}:$I${end}),0,(I{r}-MIN($I${start}:$I${end}))/(MAX($I${start}:$I${end})-MIN($I${start}:$I${end})))',
        "AQ": f'=IF(MAX($I${start}:$I${end})=0,0,I{r}/MAX($I${start}:$I${end}))',
        "AR": f'=Parameters!$C$15*AP{r}',
        "AS": f'=Parameters!$C$15+AR{r}*AQ{r}^2',
        "AT": f'=Parameters!$C$16*Parameters!$C$17*AN{r}',
        "AU": f'=Z{r}-AC{r}-AE{r}+AT{r}-AS{r}*AN{r}',
        "AV": f'=K{r}*AN{r}-Parameters!$C$3/2*AN{r}^2-AM{r}*AN{r}+Parameters!$C$16*Parameters!$C$17*AN{r}',
        "AW": f'=K{r}*AN{r}-Parameters!$C$3/2*AN{r}^2+Parameters!$C$16*Parameters!$C$17*AN{r}-K{r}*AA{r}+Parameters!$C$3/2*AA{r}^2+AC{r}+AE{r}',
        "AX": f'=IF(AN{r}=0,"",AW{r}/(2*AN{r}))',
        "AY": f'=(K{r}+Parameters!$C$16*Parameters!$C$17-AM{r})*AN{r}-Parameters!$C$3/2*AN{r}^2',
        "AZ": f'=AN{r}*AM{r}-Parameters!$C$18*AN{r}-AE{r}-Parameters!$C$4*ABS(AB{r})*AN{r}-Parameters!$C$19/2*AN{r}^2',
        "BA": f'=(K{r}+Parameters!$C$16*Parameters!$C$17-Parameters!$C$18-Parameters!$C$5-Parameters!$C$4*ABS(AB{r}))*AN{r}',
        "BB": f'=(Parameters!$C$3+Parameters!$C$19)/2*AN{r}^2',
        "BC": f'=BA{r}-BB{r}',
        "BD": f'=IF(R{r}="charge",X{r}*(K{r}+Parameters!$C$3*Parameters!$C$7-Parameters!$C$3*H{r}-L{r})-Parameters!$C$3/2*X{r}^2,IF(R{r}="discharge",X{r}*(K{r}-Parameters!$C$3*Parameters!$C$8-Parameters!$C$3*H{r}-L{r})-Parameters!$C$3/2*X{r}^2,X{r}*(K{r}-Parameters!$C$3*H{r}-L{r})-Parameters!$C$3/2*X{r}^2))',
        "BE": f'=X{r}*(K{r}-Parameters!$C$3*Parameters!$C$8-Parameters!$C$3*H{r}-L{r})-Parameters!$C$3/2*X{r}^2',
        "BF": f'=MAX($L${start}:$L${end})',
        "BG": f'=MAX(0,(BF{r}-L{r})*U{r})',
        "BH": f'=BD{r}+BG{r}',
    }


def style_title(ws, cell_range: str) -> None:
    for row in ws[cell_range]:
        for cell in row:
            cell.fill = TITLE_FILL
            cell.font = Font(color="FFFFFF", bold=True, size=14)
            cell.alignment = Alignment(horizontal="center", vertical="center")


def style_header(ws, cell_range: str) -> None:
    for row in ws[cell_range]:
        for cell in row:
            cell.fill = HEADER_FILL
            cell.font = Font(bold=True)
            cell.alignment = Alignment(horizontal="center", vertical="center", wrap_text=True)
            cell.border = Border(bottom=THIN_GREY)


def create_parameters_sheet(wb) -> None:
    if "Parameters" in wb.sheetnames:
        del wb["Parameters"]
    ws = wb.create_sheet("Parameters", 0)
    ws.merge_cells("A1:F1")
    ws["A1"] = "Model Parameters and Core Assumptions"
    style_title(ws, "A1:F1")
    ws.append(["Parameter", "Symbol", "Value", "Unit", "Used in", "Notes"])
    style_header(ws, "A2:F2")
    for item in PARAMETERS:
        ws.append(item)
    ws.freeze_panes = "A3"
    widths = {"A": 38, "B": 18, "C": 14, "D": 20, "E": 27, "F": 34}
    for col, width in widths.items():
        ws.column_dimensions[col].width = width
    for row in ws.iter_rows(min_row=3, max_row=2 + len(PARAMETERS)):
        for cell in row:
            cell.alignment = Alignment(vertical="top", wrap_text=True)
            cell.border = Border(bottom=Side(style="hair", color="D9E2F3"))
    ws["C22"].number_format = "0.0000"
    ws["C23"].number_format = "0.0000"


def create_eq_guide_sheet(wb) -> None:
    if "EQ_Guide" in wb.sheetnames:
        del wb["EQ_Guide"]
    ws = wb.create_sheet("EQ_Guide", 1)
    ws.merge_cells("A1:F1")
    ws["A1"] = "EQ1–EQ32 Formula and Output Guide"
    style_title(ws, "A1:F1")
    ws.append(["Equation", "Concept", "Symbol/output", "Excel column(s)", "Unit", "Implementation note"])
    style_header(ws, "A2:F2")
    for row in EQ_GUIDE:
        ws.append(row)
    ws.freeze_panes = "A3"
    widths = {"A": 14, "B": 28, "C": 25, "D": 18, "E": 18, "F": 58}
    for col, width in widths.items():
        ws.column_dimensions[col].width = width
    for row in ws.iter_rows(min_row=3):
        for cell in row:
            cell.alignment = Alignment(vertical="top", wrap_text=True)
            cell.border = Border(bottom=Side(style="hair", color="D9E2F3"))


def populate_hourly_sheet(ws) -> None:
    for col, title in HEADERS.items():
        ws[f"{col}1"] = title
    style_header(ws, "A1:BH1")
    ws.row_dimensions[1].height = 44
    ws.freeze_panes = "K2"
    ws.auto_filter.ref = f"A1:BH{LAST_DATA_ROW}"

    for row in range(FIRST_DATA_ROW, LAST_DATA_ROW + 1):
        formulas = formula_map(row)
        for col, formula in formulas.items():
            ws[f"{col}{row}"] = formula

    # Number formats
    for row in range(FIRST_DATA_ROW, LAST_DATA_ROW + 1):
        for col in range(11, 61):  # K:BH
            ws.cell(row=row, column=col).number_format = "0.000"
        ws[f"R{row}"].number_format = "General"
        ws[f"AX{row}"].number_format = "0.0000"

    for col in range(1, 61):
        letter = get_column_letter(col)
        if col == 1:
            ws.column_dimensions[letter].width = 19
        elif col == 2:
            ws.column_dimensions[letter].width = 13
        elif 3 <= col <= 10:
            ws.column_dimensions[letter].width = 13
        else:
            ws.column_dimensions[letter].width = 15
    ws.column_dimensions["R"].width = 20
    ws.column_dimensions["AJ"].width = 21
    ws.column_dimensions["AL"].width = 24
    ws.column_dimensions["BF"].width = 21


def create_daily_summary(wb) -> None:
    if "Daily_Summary" in wb.sheetnames:
        del wb["Daily_Summary"]
    ws = wb.create_sheet("Daily_Summary")
    ws.merge_cells("A1:N1")
    ws["A1"] = "Daily 24-hour Optimization Summary"
    style_title(ws, "A1:N1")
    headers = [
        "Day", "Date", "Rows", "Daily Utility EQ8/EQ29 (€)", "U_prosumer EQ14 (€)",
        "SW_total EQ15 (€)", "CS EQ23 (€)", "PS EQ24 (€)", "SW_P2P EQ25 (€)",
        "CS_pros EQ28 (€)", "SW_grid EQ32 (€)", "Grid purchases (kWh)",
        "P2P/Grid sales (kWh)", "Final SOC (kWh)"
    ]
    ws.append(headers)
    style_header(ws, "A2:N2")

    for day in range(DAYS_PER_YEAR):
        excel_row = day + 3
        start = FIRST_DATA_ROW + day * HOURS_PER_DAY
        end = start + HOURS_PER_DAY - 1
        ws.cell(excel_row, 1, day + 1)
        ws.cell(excel_row, 2, f"='{DATA_SHEET}'!B{start}")
        ws.cell(excel_row, 3, f'="{start}–{end}"')
        formulas = [
            f"=SUM('{DATA_SHEET}'!AI{start}:AI{end})",
            f"=SUM('{DATA_SHEET}'!AO{start}:AO{end})",
            f"=SUM('{DATA_SHEET}'!AU{start}:AU{end})",
            f"=SUM('{DATA_SHEET}'!AY{start}:AY{end})",
            f"=SUM('{DATA_SHEET}'!AZ{start}:AZ{end})",
            f"=SUM('{DATA_SHEET}'!BC{start}:BC{end})",
            f"=SUM('{DATA_SHEET}'!BD{start}:BD{end})",
            f"=SUM('{DATA_SHEET}'!BH{start}:BH{end})",
            f"=SUM('{DATA_SHEET}'!AG{start}:AG{end})",
            f"=SUM('{DATA_SHEET}'!AF{start}:AF{end})",
            f"='{DATA_SHEET}'!W{end}",
        ]
        for col, formula in enumerate(formulas, start=4):
            ws.cell(excel_row, col, formula)

    ws.freeze_panes = "A3"
    ws.auto_filter.ref = f"A2:N{DAYS_PER_YEAR + 2}"
    ws.column_dimensions["A"].width = 9
    ws.column_dimensions["B"].width = 13
    ws.column_dimensions["C"].width = 13
    for col in range(4, 15):
        ws.column_dimensions[get_column_letter(col)].width = 22
    for row in range(3, DAYS_PER_YEAR + 3):
        ws[f"B{row}"].number_format = "yyyy-mm-dd"
        for col in range(4, 15):
            ws.cell(row, col).number_format = "#,##0.000"


def create_monthly_summary(wb) -> None:
    if "Monthly_Summary" in wb.sheetnames:
        del wb["Monthly_Summary"]
    ws = wb.create_sheet("Monthly_Summary")
    ws.merge_cells("A1:K1")
    ws["A1"] = "Monthly Aggregation"
    style_title(ws, "A1:K1")
    headers = ["Month", "Utility EQ8/EQ29", "U_prosumer EQ14", "SW_total EQ15", "CS EQ23", "PS EQ24", "SW_P2P EQ25", "CS_pros EQ28", "SW_grid EQ32", "E_buy", "E_sell"]
    ws.append(headers)
    style_header(ws, "A2:K2")
    for month in range(1, 13):
        row = month + 2
        ws.cell(row, 1, month)
        criteria = f'MONTH(\'{DATA_SHEET}\'!$B$2:$B${LAST_DATA_ROW})={month}'
        source_cols = ["AI", "AO", "AU", "AY", "AZ", "BC", "BD", "BH", "AG", "AF"]
        for idx, src in enumerate(source_cols, start=2):
            ws.cell(row, idx, f'=SUMPRODUCT(--({criteria}),\'{DATA_SHEET}\'!${src}$2:${src}${LAST_DATA_ROW})')
    ws.freeze_panes = "A3"
    ws.column_dimensions["A"].width = 12
    for col in range(2, 12):
        ws.column_dimensions[get_column_letter(col)].width = 20
        for row in range(3, 15):
            ws.cell(row, col).number_format = "#,##0.000"


def create_annual_summary(wb) -> None:
    if "Annual_Summary" in wb.sheetnames:
        del wb["Annual_Summary"]
    ws = wb.create_sheet("Annual_Summary")
    ws.merge_cells("A1:C1")
    ws["A1"] = "Annual Model KPIs"
    style_title(ws, "A1:C1")
    ws.append(["KPI", "Value", "Unit"])
    style_header(ws, "A2:C2")
    rows = [
        ("Annual Utility EQ8/EQ29", f"=SUM('{DATA_SHEET}'!AI2:AI{LAST_DATA_ROW})", "€"),
        ("Annual U_prosumer EQ14", f"=SUM('{DATA_SHEET}'!AO2:AO{LAST_DATA_ROW})", "€"),
        ("Annual SW_total EQ15", f"=SUM('{DATA_SHEET}'!AU2:AU{LAST_DATA_ROW})", "€"),
        ("Annual CS EQ23", f"=SUM('{DATA_SHEET}'!AY2:AY{LAST_DATA_ROW})", "€"),
        ("Annual PS EQ24", f"=SUM('{DATA_SHEET}'!AZ2:AZ{LAST_DATA_ROW})", "€"),
        ("Annual SW_P2P EQ25", f"=SUM('{DATA_SHEET}'!BC2:BC{LAST_DATA_ROW})", "€"),
        ("Annual CS_pros EQ28", f"=SUM('{DATA_SHEET}'!BD2:BD{LAST_DATA_ROW})", "€"),
        ("Annual SW_grid EQ32", f"=SUM('{DATA_SHEET}'!BH2:BH{LAST_DATA_ROW})", "€"),
        ("Annual grid purchases", f"=SUM('{DATA_SHEET}'!AG2:AG{LAST_DATA_ROW})", "kWh"),
        ("Annual exported/sold energy", f"=SUM('{DATA_SHEET}'!AF2:AF{LAST_DATA_ROW})", "kWh"),
        ("Annual battery charging", f"=SUM('{DATA_SHEET}'!T2:T{LAST_DATA_ROW})", "kWh"),
        ("Annual battery discharging", f"=SUM('{DATA_SHEET}'!U2:U{LAST_DATA_ROW})", "kWh"),
    ]
    for item in rows:
        ws.append(item)
    ws.column_dimensions["A"].width = 36
    ws.column_dimensions["B"].width = 20
    ws.column_dimensions["C"].width = 16
    for row in range(3, 3 + len(rows)):
        ws[f"B{row}"].number_format = "#,##0.000"


def add_readme_sheet(wb) -> None:
    if "README" in wb.sheetnames:
        del wb["README"]
    ws = wb.create_sheet("README", 0)
    ws.merge_cells("A1:F1")
    ws["A1"] = "EQ1–EQ32 Daily Optimization Workbook"
    style_title(ws, "A1:F1")
    content = [
        ("Purpose", "Formula-driven residential prosumer/P2P electricity-market workbook for 8,760 hourly observations."),
        ("Optimization horizon", "365 fixed daily blocks of 24 hours. The model does not use a rolling/sliding window."),
        ("Formula visibility", "Calculated cells contain native Excel formulas visible in the Formula Bar."),
        ("Decision labels", "Buy from the Grid, discharge, charge, sell, self consume."),
        ("E_sell", "PV minus self-consumption; when charging, charging energy is also subtracted."),
        ("E_buy", "MAX(0, baseline demand − self-consumption − battery discharge)."),
        ("Recalculation", "Open the workbook in Microsoft Excel and allow calculation to refresh all formula results."),
        ("Generator", "build_EQ1_EQ32_DailyOptimization.py, using openpyxl."),
    ]
    ws["A3"] = "Item"
    ws["B3"] = "Description"
    style_header(ws, "A3:B3")
    for i, (item, desc) in enumerate(content, start=4):
        ws[f"A{i}"] = item
        ws[f"B{i}"] = desc
        ws[f"A{i}"].font = Font(bold=True)
        ws[f"A{i}"].alignment = Alignment(vertical="top", wrap_text=True)
        ws[f"B{i}"].alignment = Alignment(vertical="top", wrap_text=True)
    ws.column_dimensions["A"].width = 26
    ws.column_dimensions["B"].width = 95


def main() -> None:
    if not INPUT_DATA_FILE.exists():
        raise FileNotFoundError(
            f"Input workbook not found: {INPUT_DATA_FILE}\n"
            "Place Input Data(2).xlsx in the same directory as this script."
        )

    wb = load_workbook(INPUT_DATA_FILE)
    if DATA_SHEET not in wb.sheetnames:
        raise KeyError(f"Worksheet {DATA_SHEET!r} was not found in {INPUT_DATA_FILE.name}")

    # Ask Excel to fully recalculate formulas when the workbook opens.
    try:
        wb.calculation.fullCalcOnLoad = True
        wb.calculation.forceFullCalc = True
        wb.calculation.calcMode = "auto"
    except AttributeError:
        pass

    create_parameters_sheet(wb)
    create_eq_guide_sheet(wb)
    populate_hourly_sheet(wb[DATA_SHEET])
    create_daily_summary(wb)
    create_monthly_summary(wb)
    create_annual_summary(wb)
    add_readme_sheet(wb)

    wb.save(OUTPUT_FILE)
    print(f"Created: {OUTPUT_FILE}")
    print(f"Hourly rows: {FIRST_DATA_ROW}–{LAST_DATA_ROW}")
    print("Optimization horizon: 365 fixed daily 24-hour periods")


if __name__ == "__main__":
    main()
```
