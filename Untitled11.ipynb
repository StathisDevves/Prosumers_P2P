```python
import sys
!{sys.executable} -m pip install artifact-tool

"""
Build the EQ1–EQ32 rolling-24-hour prosumer/P2P Excel model.

Requirements
------------
- Python 3.10+
- artifact_tool
- Input workbook: Input Data.xlsx

The script preserves the original hourly input sheet, adds formula-driven
columns K:BH, and creates Parameters, EQ_Guide, and Summary worksheets.
All calculated cells contain native Excel formulas.
"""

from pathlib import Path
from artifact_tool import Blob, SpreadsheetFile

INPUT_FILE = Path("/mnt/data/Input Data.xlsx")
OUTPUT_FILE = Path("/mnt/data/Output_EQ1_EQ32_Rolling24h_from_Python.xlsx")
DATA_SHEET = "Solar_8760_2025"
FIRST_DATA_ROW = 2
LAST_DATA_ROW = 8761


def main() -> None:
    wb = SpreadsheetFile.import_xlsx(Blob.load(str(INPUT_FILE)))
    if wb is None:
        raise RuntimeError(f"Could not import {INPUT_FILE}")

    ws = wb.worksheets.get_item(DATA_SHEET)

    # ------------------------------------------------------------------
    # Parameters
    # ------------------------------------------------------------------
    p = wb.worksheets.get_or_add("Parameters")
    p.get_range("A1:F25").clear({})
    p.merge_cells("A1:F1")
    p.get_range("A1").values = [["Model Parameters and Core Assumptions"]]
    p.get_range("A2:F2").values = [[
        "Parameter", "Symbol", "Value", "Unit", "Used in", "Notes"
    ]]
    parameter_rows = [
        ["Utility curvature", "alpha", 0.25, "€/kWh²", "EQ1–EQ8, EQ19–EQ30", "User-selected"],
        ["Load-shifting coefficient", "zeta", 0.20, "€/kWh²", "EQ8, EQ14, EQ21, EQ24–EQ29", "User-selected"],
        ["Battery degradation coefficient", "psi", 0.03, "€/kWh", "EQ10, EQ14, EQ21, EQ24–EQ29", "User-selected"],
        ["Battery capacity", "C_bat", 30.0, "kWh", "EQ2, EQ4, EQ9, EQ12", "Core assumption"],
        ["Maximum charging power", "P_ch_max", 3.0, "kWh/h", "EQ2, EQ11", "One-hour step"],
        ["Maximum discharging power", "P_dis_max", 3.0, "kWh/h", "EQ4, EQ11, EQ28–EQ31", "One-hour step"],
        ["Charging efficiency", "eta_ch", 0.92, "fraction", "EQ9, EQ12", "Core assumption"],
        ["Discharging efficiency", "eta_dis", 0.92, "fraction", "EQ9, EQ12", "One-way proxy"],
        ["Initial SOC", "SOC_0", 0.90, "fraction of capacity", "EQ9, EQ12", "90%"],
        ["Minimum SOC", "SOC_min", 0.10, "fraction of capacity", "EQ4, EQ9, EQ12", "Operational bound"],
        ["Maximum SOC", "SOC_max", 1.00, "fraction of capacity", "EQ2, EQ9, EQ12", "Operational bound"],
        ["Local supply-demand ratio", "SDR", 0.50, "fraction", "EQ13", "Given example"],
        ["Baseline network cost", "tau_base", 0.05, "€/kWh", "EQ17", "Lower stated bound"],
        ["Green preference premium", "theta", 0.022, "€/kWh", "EQ15, EQ19, EQ21, EQ23–EQ27", "Given"],
        ["Green guarantee indicator", "rho", 1.00, "fraction", "EQ15, EQ19, EQ21, EQ23–EQ27", "1 = fully green"],
        ["Prosumer marginal PV cost", "b_i", 0.065, "€/kWh", "EQ24–EQ27", "Range midpoint"],
        ["Prosumer quadratic cost", "a_i", 0.025, "€/kWh²", "EQ24–EQ27", "Range midpoint"],
        ["Network scaling coefficient", "a1_min", 0.10, "fraction", "EQ18", "At minimum load"],
        ["Network scaling coefficient", "a1_max", 0.50, "fraction", "EQ18", "At maximum load"],
        ["Low-price threshold share", "Low_share", 1 / 3, "fraction", "EQ2 rule", "Lower third of rolling range"],
        ["High-price threshold share", "High_share", 2 / 3, "fraction", "EQ4 rule", "Upper third of rolling range"],
    ]
    p.get_range("A3:F23").values = parameter_rows

    # ------------------------------------------------------------------
    # Main output headers
    # ------------------------------------------------------------------
    headers = [
        "ω_t (€/kWh)", "Price_spot (€/kWh)", "Roll24 Price Min", "Roll24 Price Max",
        "Low Threshold", "High Threshold", "Roll24 PV Max", "Decision",
        "E_self EQ1 (kWh)", "E_charge (kWh)", "E_discharge (kWh)",
        "SOC_begin (kWh)", "SOC_end EQ9/12 (kWh)", "E_grid EQ2-4 (kWh)",
        "Elasticity EQ6", "U_cons EQ7 (€)", "E_actual EQ22 (kWh)", "DeltaE (kWh)",
        "D_shift (€)", "E_bat EQ11 (kWh)", "C_deg EQ10 (€)", "E_sell (kWh)",
        "E_buy (kWh)", "W_spot (€)", "Utility EQ8/29 (€)", "Roll24 Umax EQ8/29 (€)",
        "SDR", "Price_prosumer (€/kWh)", "Price_P2P EQ13 (€/kWh)",
        "E_inj EQ16 (kWh)", "U_prosumer EQ14 (€)", "a1 EQ18", "Ltotal/Cgrid",
        "gamma EQ18", "tau EQ17 (€/kWh)", "Green benefit (€)", "SW_total EQ15 (€)",
        "U_buyer EQ19 (€)", "A EQ21 (€)", "Price_eq EQ20 (€/kWh)", "CS EQ23 (€)",
        "PS EQ24 (€)", "LC EQ26 (€)", "QC EQ27 (€)", "SW_P2P EQ25 (€)",
        "CS_pros EQ28 (€)", "CSt1 EQ30 (€)", "Price_peak24 (€/kWh)",
        "GS_grid EQ31 (€)", "SW_grid EQ32 (€)"
    ]
    ws.get_range("K1:BH1").values = [headers]

    # Formula seed for row 2. Filling down preserves relative references.
    formula_by_col = {
        "K": '=IF(OR(F2<=6,F2=24),0.55,IF(F2<=10,1,IF(F2<=16,0.8,IF(F2<=22,1.35,0.65))))',
        "L": '=J2/1000',
        "M": '=MIN(L2:L25)',
        "N": '=MAX(L2:L25)',
        "O": '=M2+(N2-M2)*Parameters!$C$22',
        "P": '=M2+(N2-M2)*Parameters!$C$23',
        "Q": '=MAX(H2:H25)',
        "R": '=IF(AND(L2<=O2,H2>S2,V2<Parameters!$C$6*Parameters!$C$13),"CHARGE",IF(AND(L2>=P2,V2>Parameters!$C$6*Parameters!$C$12),"DISCHARGE",IF(H2>=S2,"SELF-CONSUME","GRID/STANDBY")))',
        "S": '=MAX(0,MIN(I2,(K2-L2)/Parameters!$C$3))',
        "T": '=IF(R2="CHARGE",MAX(0,MIN(Parameters!$C$7,Parameters!$C$6*Parameters!$C$13-V2,MAX(0,H2-S2))),0)',
        "U": '=IF(R2="DISCHARGE",MAX(0,MIN(Parameters!$C$8,(V2-Parameters!$C$6*Parameters!$C$12)*Parameters!$C$10,MAX(0,S2-H2))),0)',
        "V": '=Parameters!$C$6*Parameters!$C$11',
        "W": '=MAX(Parameters!$C$6*Parameters!$C$12,MIN(Parameters!$C$6*Parameters!$C$13,V2+Parameters!$C$9*T2-U2/Parameters!$C$10))',
        "X": '=MAX(0,S2+T2-U2-H2)',
        "Y": '=IF(X2=0,0,-(1/Parameters!$C$3)*L2/X2)',
        "Z": '=K2*S2-Parameters!$C$3/2*S2^2',
        "AA": '=AG2+S2',
        "AB": '=I2-AA2',
        "AC": '=Parameters!$C$4/2*AB2^2',
        "AD": '=T2-U2',
        "AE": '=Parameters!$C$5*ABS(AD2)',
        "AF": '=MAX(0,H2-S2-T2)',
        "AG": '=MAX(0,I2-S2+U2)',
        "AH": '=L2*(AF2-AG2)',
        "AI": '=Z2-AC2+AH2-AE2',
        "AJ": '=SUM(AI2:AI25)',
        "AK": '=IF(AG2=0,IF(AF2>0,1,0),MIN(1,AF2/AG2))',
        "AL": '=IFERROR(INDEX(L2:L25,MATCH(Q2,H2:H25,0)),N2)',
        "AM": '=IFERROR(L2*AL2/(L2-AL2)*Parameters!$C$14+AL2,(L2+AL2)/2)',
        "AN": '=MAX(0,H2-S2-T2)',
        "AO": '=Z2-AC2+AN2*AM2-AE2',
        "AP": '=Parameters!$C$20+(Parameters!$C$21-Parameters!$C$20)*IF(MAX(I2:I25)=MIN(I2:I25),0,(I2-MIN(I2:I25))/(MAX(I2:I25)-MIN(I2:I25)))',
        "AQ": '=IF(MAX(I2:I25)=0,0,I2/MAX(I2:I25))',
        "AR": '=Parameters!$C$15*AP2',
        "AS": '=Parameters!$C$15+AR2*AQ2^2',
        "AT": '=Parameters!$C$16*Parameters!$C$17*AN2',
        "AU": '=Z2-AC2-AE2+AT2-AS2*AN2',
        "AV": '=K2*AN2-Parameters!$C$3/2*AN2^2-AM2*AN2+Parameters!$C$16*Parameters!$C$17*AN2',
        "AW": '=K2*AN2-Parameters!$C$3/2*AN2^2+Parameters!$C$16*Parameters!$C$17*AN2-K2*AA2+Parameters!$C$3/2*AA2^2+AC2+AE2',
        "AX": '=IF(AN2=0,"",AW2/(2*AN2))',
        "AY": '=(K2+Parameters!$C$16*Parameters!$C$17-AM2)*AN2-Parameters!$C$3/2*AN2^2',
        "AZ": '=AN2*AM2-Parameters!$C$18*AN2-AE2-Parameters!$C$4*ABS(AB2)*AN2-Parameters!$C$19/2*AN2^2',
        "BA": '=(K2+Parameters!$C$16*Parameters!$C$17-Parameters!$C$18-Parameters!$C$5-Parameters!$C$4*ABS(AB2))*AN2',
        "BB": '=(Parameters!$C$3+Parameters!$C$19)/2*AN2^2',
        "BC": '=BA2-BB2',
        "BD": '=IF(R2="CHARGE",X2*(K2+Parameters!$C$3*Parameters!$C$7-Parameters!$C$3*H2-L2)-Parameters!$C$3/2*X2^2,IF(R2="DISCHARGE",X2*(K2-Parameters!$C$3*Parameters!$C$8-Parameters!$C$3*H2-L2)-Parameters!$C$3/2*X2^2,X2*(K2-Parameters!$C$3*H2-L2)-Parameters!$C$3/2*X2^2))',
        "BE": '=X2*(K2-Parameters!$C$3*Parameters!$C$8-Parameters!$C$3*H2-L2)-Parameters!$C$3/2*X2^2',
        "BF": '=MAX(L2:L25)',
        "BG": '=MAX(0,(BF2-L2)*U2)',
        "BH": '=BD2+BG2',
    }
    cols = list(formula_by_col)
    ws.get_range("K2:BH2").formulas = [[formula_by_col[c] for c in cols]]
    ws.get_range(f"K2:BH{LAST_DATA_ROW}").fill_down()

    # Carry SOC from previous hour after the first row.
    ws.get_range("V3").formulas = [["=W2"]]
    ws.get_range(f"V3:V{LAST_DATA_ROW}").fill_down()

    # ------------------------------------------------------------------
    # Daily and annual aggregation
    # ------------------------------------------------------------------
    s = wb.worksheets.get_or_add("Summary")
    s.get_range("A1:L370").clear({})
    s.merge_cells("A1:I1")
    s.get_range("A1").values = [["Rolling-24h Utility and Welfare Aggregations"]]
    s.get_range("A2:I2").values = [[
        "Date", "Daily Umax EQ8/29", "Daily Uprosumer EQ14", "Daily SWtotal EQ15",
        "Daily CS EQ23", "Daily PS EQ24", "Daily SWP2P EQ25", "Daily CSpros EQ28",
        "Daily SWgrid EQ32"
    ]]
    daily = []
    for day in range(365):
        r0 = FIRST_DATA_ROW + day * 24
        daily.append([
            f"='{DATA_SHEET}'!B{r0}",
            f"=SUM('{DATA_SHEET}'!AI{r0}:AI{r0+23})",
            f"=SUM('{DATA_SHEET}'!AO{r0}:AO{r0+23})",
            f"=SUM('{DATA_SHEET}'!AU{r0}:AU{r0+23})",
            f"=SUM('{DATA_SHEET}'!AY{r0}:AY{r0+23})",
            f"=SUM('{DATA_SHEET}'!AZ{r0}:AZ{r0+23})",
            f"=SUM('{DATA_SHEET}'!BC{r0}:BC{r0+23})",
            f"=SUM('{DATA_SHEET}'!BD{r0}:BD{r0+23})",
            f"=SUM('{DATA_SHEET}'!BH{r0}:BH{r0+23})",
        ])
    s.get_range("A3:I367").formulas = daily
    s.get_range("K2:L2").values = [["Annual KPI", "Value (€)"]]
    labels = [
        "Annual Umax EQ8/29", "Annual Uprosumer EQ14", "Annual SWtotal EQ15",
        "Annual CS EQ23", "Annual PS EQ24", "Annual SWP2P EQ25",
        "Annual CSpros EQ28", "Annual SWgrid EQ32"
    ]
    s.get_range("K3:K10").values = [[x] for x in labels]
    s.get_range("L3:L10").formulas = [[f"=SUM({c}3:{c}367)"] for c in "BCDEFGHI"]

    # ------------------------------------------------------------------
    # Formatting
    # ------------------------------------------------------------------
    title_fmt = {
        "fill": "#1F4E78",
        "font": {"bold": True, "color": "#FFFFFF", "size": 14},
        "horizontal_alignment": "center",
        "vertical_alignment": "center",
    }
    header_fmt = {
        "fill": "#D9EAF7",
        "font": {"bold": True},
        "horizontal_alignment": "center",
        "vertical_alignment": "center",
        "wrap_text": True,
    }
    p.get_range("A1:F1").format = title_fmt
    p.get_range("A2:F2").format = header_fmt
    s.get_range("A1:I1").format = title_fmt
    s.get_range("A2:I2").format = header_fmt
    s.get_range("K2:L2").format = header_fmt
    ws.get_range("A1:BH1").format = header_fmt

    ws.freeze_panes.freeze_rows(1)
    ws.freeze_panes.freeze_columns(10)
    p.freeze_panes.freeze_rows(2)
    s.freeze_panes.freeze_rows(2)

    ws.get_range(f"A2:A{LAST_DATA_ROW}").format.number_format = "yyyy-mm-dd hh:mm"
    ws.get_range(f"B2:B{LAST_DATA_ROW}").format.number_format = "yyyy-mm-dd"
    ws.get_range(f"K2:BH{LAST_DATA_ROW}").format.number_format = "0.000"
    s.get_range("A3:A367").format.number_format = "yyyy-mm-dd"
    s.get_range("B3:I367").format.number_format = "#,##0.00"
    s.get_range("L3:L10").format.number_format = "#,##0.00"

    ws.get_range("A:A").format.column_width = 18
    ws.get_range("B:B").format.column_width = 12
    ws.get_range("C:J").format.column_width = 13
    ws.get_range("K:BH").format.column_width = 14
    ws.get_range("R:R").format.column_width = 18
    ws.get_range("A1:BH1").format.row_height = 42
    p.get_range("A:E").format.column_width = 20
    p.get_range("F:F").format.column_width = 32
    s.get_range("A:I").format.column_width = 18
    s.get_range("K:K").format.column_width = 28
    s.get_range("L:L").format.column_width = 16

    SpreadsheetFile.export_xlsx(wb).save(str(OUTPUT_FILE))
    print(f"Created: {OUTPUT_FILE}")


if __name__ == "__main__":
    main()
```

### Ορισμός Παραμέτρων Μοντέλου

Στη συνέχεια, θα ορίσουμε τις παραμέτρους του μοντέλου που χρησιμοποιούνται για τους υπολογισμούς. Αυτές οι τιμές προέρχονται από το αρχικό φύλλο "Parameters" του Excel και θα αποθηκευτούν σε ένα λεξικό Python για εύκολη πρόσβαση.

```python
# Ορισμός παραμέτρων σε ένα λεξικό
parameters = {
    "alpha": 0.25,  # Utility curvature (€/kWh²)
    "zeta": 0.20,   # Load-shifting coefficient (€/kWh²)
    "psi": 0.03,    # Battery degradation coefficient (€/kWh)
    "C_bat": 30.0,  # Battery capacity (kWh)
    "P_ch_max": 3.0, # Maximum charging power (kWh/h)
    "P_dis_max": 3.0, # Maximum discharging power (kWh/h)
    "eta_ch": 0.92, # Charging efficiency (fraction)
    "eta_dis": 0.92, # Discharging efficiency (fraction)
    "SOC_0": 0.90,  # Initial SOC (fraction of capacity)
    "SOC_min": 0.10, # Minimum SOC (fraction of capacity)
    "SOC_max": 1.00, # Maximum SOC (fraction of capacity)
    "SDR": 0.50,    # Local supply-demand ratio (fraction)
    "tau_base": 0.05, # Baseline network cost (€/kWh)
    "theta": 0.022, # Green preference premium (€/kWh)
    "rho": 1.00,    # Green guarantee indicator (fraction)
    "b_i": 0.065,   # Prosumer marginal PV cost (€/kWh)
    "a_i": 0.025,   # Prosumer quadratic cost (€/kWh²)
    "a1_min": 0.10, # Network scaling coefficient (at minimum load)
    "a1_max": 0.50, # Network scaling coefficient (at maximum load)
    "Low_share": 1 / 3, # Low-price threshold share (fraction)
    "High_share": 2 / 3 # High-price threshold share (fraction)
}

print("Οι παράμετροι του μοντέλου ορίστηκαν επιτυχώς:")
for key, value in parameters.items():
    print(f"  {key}: {value}")
```

### Υπολογισμός Νέων Στηλών

Τώρα θα προσθέσουμε τις νέες στήλες στο DataFrame `df_excel`, μεταφράζοντας τους αρχικούς τύπους του Excel σε αντίστοιχες λειτουργίες του `pandas`. Θα ξεκινήσουμε με τις πρώτες στήλες που ορίζουν την τιμή και τα όρια.

```python
# Στήλη K: ω_t (€/kWh) - Προσαρμοσμένη τιμή με βάση την ώρα
def calculate_omega_t(hour):
    if hour <= 6 or hour == 24:
        return 0.55
    elif hour <= 10:
        return 1.0
    elif hour <= 16:
        return 0.8
    elif hour <= 22:
        return 1.35
    else: # hour == 23
        return 0.65

df_excel['ω_t (€/kWh)'] = df_excel['Hour'].apply(calculate_omega_t)

# Στήλη L: Price_spot (€/kWh) - Σποτ τιμή σε €/kWh
df_excel['Price_spot (€/kWh)'] = df_excel['Price,spot,t (€/MWh)'] / 1000

# Στήλη M: Roll24 Price Min - Κυλιόμενο ελάχιστο τιμής σποτ 24 ωρών
df_excel['Roll24 Price Min'] = df_excel['Price_spot (€/kWh)'].rolling(window=24, min_periods=1).min()

# Στήλη N: Roll24 Price Max - Κυλιόμενο μέγιστο τιμής σποτ 24 ωρών
df_excel['Roll24 Price Max'] = df_excel['Price_spot (€/kWh)'].rolling(window=24, min_periods=1).max()

# Στήλη O: Low Threshold - Κατώτερο όριο τιμής
df_excel['Low Threshold'] = df_excel['Roll24 Price Min'] + (df_excel['Roll24 Price Max'] - df_excel['Roll24 Price Min']) * parameters['Low_share']

# Στήλη P: High Threshold - Ανώτερο όριο τιμής
df_excel['High Threshold'] = df_excel['Roll24 Price Min'] + (df_excel['Roll24 Price Max'] - df_excel['Roll24 Price Min']) * parameters['High_share']

# Στήλη Q: Roll24 PV Max - Κυλιόμενο μέγιστο παραγωγής PV 24 ωρών
df_excel['Roll24 PV Max'] = df_excel['Epv _PV,Generation (KWh)'].rolling(window=24, min_periods=1).max()

print("Οι στήλες K έως Q υπολογίστηκαν επιτυχώς. Δείτε τις πρώτες γραμμές του ενημερωμένου DataFrame:")
display(df_excel[['Hour', 'ω_t (€/kWh)', 'Price_spot (€/kWh)', 'Roll24 Price Min', 'Roll24 Price Max', 'Low Threshold', 'High Threshold', 'Roll24 PV Max']].head(25))
```

### Υπολογισμός Στηλών R έως AH

Συνεχίζουμε την προσθήκη υπολογιζόμενων στηλών στο DataFrame, μεταφράζοντας τους επόμενους τύπους του Excel σε `pandas` operations.

```python
import numpy as np

# Στήλη S: E_self EQ1 (kWh) = MAX(0,MIN(I2,(K2-L2)/Parameters!$C$3))
# Calculate E_self EQ1 (kWh) for all rows (this is not iteratively dependent)
df_excel['E_self EQ1 (kWh)'] = np.maximum(0,
                                         np.minimum(df_excel['E,base,t Base load Demand_kWh'],
                                                    (df_excel['ω_t (€/kWh)'] - df_excel['Price_spot (€/kWh)']) / parameters['alpha']))

# Initialize iterative columns with default values
df_excel['Decision'] = ''
df_excel['E_charge (kWh)'] = 0.0
df_excel['E_discharge (kWh)'] = 0.0
df_excel['SOC_begin (kWh)'] = 0.0  # Will be updated in loop
df_excel['SOC_end EQ9/12 (kWh)'] = 0.0 # Will be updated in loop
df_excel['E_grid EQ2-4 (kWh)'] = 0.0 # No formula provided, initializing to 0.0

# Loop through each hour to handle time-series dependencies for SOC and decisions
for i in range(len(df_excel)):
    # V: SOC_begin (kWh) - Carry SOC from previous hour, or use initial SOC_0
    if i == 0:
        current_soc_begin = parameters['C_bat'] * parameters['SOC_0']
    else:
        current_soc_begin = df_excel.loc[i-1, 'SOC_end EQ9/12 (kWh)']

    df_excel.loc[i, 'SOC_begin (kWh)'] = current_soc_begin

    pv_gen = df_excel.loc[i, 'Epv _PV,Generation (KWh)']
    base_load_demand = df_excel.loc[i, 'E,base,t Base load Demand_kWh']
    spot_price = df_excel.loc[i, 'Price_spot (€/kWh)']
    low_threshold = df_excel.loc[i, 'Low Threshold']
    high_threshold = df_excel.loc[i, 'High Threshold']
    e_self_eq1 = df_excel.loc[i, 'E_self EQ1 (kWh)'] # Already calculated vectorized

    decision = "GRID/STANDBY" # Default decision
    e_charge = 0.0
    e_discharge = 0.0

    # R: Decision - Logic based on Excel formula
    # CHARGE condition: L2<=O2, H2>S2, V2<C_bat*SOC_max
    if (spot_price <= low_threshold) and \
       (pv_gen > e_self_eq1) and \
       (current_soc_begin < parameters['C_bat'] * parameters['SOC_max']):
        decision = "CHARGE"
        # T: E_charge (kWh)
        e_charge = np.maximum(0, np.minimum(
            parameters['P_ch_max'],
            np.minimum(parameters['C_bat'] * parameters['SOC_max'] - current_soc_begin,
                       np.maximum(0, pv_gen - e_self_eq1))
        ))
    # DISCHARGE condition: L2>=P2, V2>C_bat*SOC_min
    elif (spot_price >= high_threshold) and \
         (current_soc_begin > parameters['C_bat'] * parameters['SOC_min']):
        decision = "DISCHARGE"
        # U: E_discharge (kWh)
        e_discharge = np.maximum(0, np.minimum(
            parameters['P_dis_max'],
            np.minimum((current_soc_begin - parameters['C_bat'] * parameters['SOC_min']) * parameters['eta_dis'],
                       np.maximum(0, e_self_eq1 - pv_gen))
        ))
    # SELF-CONSUME condition: H2>=S2
    elif pv_gen >= e_self_eq1:
        decision = "SELF-CONSUME"

    df_excel.loc[i, 'Decision'] = decision
    df_excel.loc[i, 'E_charge (kWh)'] = e_charge
    df_excel.loc[i, 'E_discharge (kWh)'] = e_discharge

    # W: SOC_end EQ9/12 (kWh)
    soc_end = np.maximum(
        parameters['C_bat'] * parameters['SOC_min'],
        np.minimum(
            parameters['C_bat'] * parameters['SOC_max'],
            current_soc_begin + parameters['eta_ch'] * e_charge - e_discharge / parameters['eta_dis']
        )
    )
    df_excel.loc[i, 'SOC_end EQ9/12 (kWh)'] = soc_end


# Now calculate the remaining columns vectorially as their dependencies are met

# X: DeltaE (kWh) = MAX(0,S2+T2-U2-H2) -> MAX(0, E_self + E_charge - E_discharge - PV_Gen)
df_excel['DeltaE (kWh)'] = np.maximum(0,
                                       df_excel['E_self EQ1 (kWh)'] +
                                       df_excel['E_charge (kWh)'] -
                                       df_excel['E_discharge (kWh)'] -
                                       df_excel['Epv _PV,Generation (KWh)'])

# Y: Elasticity EQ6 = IF(X2=0,0,-(1/Parameters!$C$3)*L2/X2)
# Note: Parameters!$C$3 is 'alpha'
df_excel['Elasticity EQ6'] = np.where(df_excel['DeltaE (kWh)'] == 0, 0,
                                       -(1/parameters['alpha']) * df_excel['Price_spot (€/kWh)'] / df_excel['DeltaE (kWh)'])

# Z: U_cons EQ7 (€) = K2*S2-Parameters!$C$3/2*S2^2
# Note: K2 is 'ω_t (€/kWh)', S2 is 'E_self EQ1 (kWh)', Parameters!$C$3 is 'alpha'
df_excel['U_cons EQ7 (€)'] = df_excel['ω_t (€/kWh)'] * df_excel['E_self EQ1 (kWh)'] - \
                            parameters['alpha'] / 2 * df_excel['E_self EQ1 (kWh)']**2

# AF: E_sell (kWh) = MAX(0, H2-S2-T2) -> MAX(0, PV_Gen - E_self - E_charge)
df_excel['E_sell (kWh)'] = np.maximum(0,
                                      df_excel['Epv _PV,Generation (KWh)'] -
                                      df_excel['E_self EQ1 (kWh)'] -
                                      df_excel['E_charge (kWh)'])

# AG: E_buy (kWh) = MAX(0, I2-S2+U2) -> MAX(0, Base_load_Demand - E_self + E_discharge)
df_excel['E_buy (kWh)'] = np.maximum(0,
                                     df_excel['E,base,t Base load Demand_kWh'] -
                                     df_excel['E_self EQ1 (kWh)'] +
                                     df_excel['E_discharge (kWh)'])

# AA: E_actual EQ22 (kWh) = AG2+S2
df_excel['E_actual EQ22 (kWh)'] = df_excel['E_buy (kWh)'] + df_excel['E_self EQ1 (kWh)']

# AB equivalent: Intermediate calculation for D_shift (€)
# This was `I2-AA2` in Excel. Let's use a temporary variable to avoid overwriting `DeltaE (kWh)`
delta_e_ab_calc = df_excel['E,base,t Base load Demand_kWh'] - df_excel['E_actual EQ22 (kWh)']

# AC: D_shift (€) = Parameters!$C$4/2*AB2^2
# Note: Parameters!$C$4 is 'zeta'
df_excel['D_shift (€)'] = parameters['zeta'] / 2 * delta_e_ab_calc**2

# AD: E_bat EQ11 (kWh) = T2-U2
df_excel['E_bat EQ11 (kWh)'] = df_excel['E_charge (kWh)'] - df_excel['E_discharge (kWh)']

# AE: C_deg EQ10 (€) = Parameters!$C$5*ABS(AD2)
# Note: Parameters!$C$5 is 'psi'
df_excel['C_deg EQ10 (€)'] = parameters['psi'] * abs(df_excel['E_bat EQ11 (kWh)'])

# AH: W_spot (€) = L2*(AF2-AG2)
# Note: L2 is 'Price_spot (€/kWh)', AF2 is 'E_sell (kWh)', AG2 is 'E_buy (kWh)'
df_excel['W_spot (€)'] = df_excel['Price_spot (€/kWh)'] * (df_excel['E_sell (kWh)'] - df_excel['E_buy (kWh)'])

# AI: Utility EQ8/29 (€) = Z2-AC2+AH2-AE2
# Z2 is 'U_cons EQ7 (€)', AC2 is 'D_shift (€)', AH2 is 'W_spot (€)', AE2 is 'C_deg EQ10 (€)'
df_excel['Utility EQ8/29 (€)'] = df_excel['U_cons EQ7 (€)'] - \
                                df_excel['D_shift (€)'] + \
                                df_excel['W_spot (€)'] - \
                                df_excel['C_deg EQ10 (€)']

print("Οι στήλες R έως AI υπολογίστηκαν επιτυχώς. Δείτε τις πρώτες γραμμές του ενημερωμένου DataFrame:")
display(df_excel[[
    'Hour', 'Decision', 'E_self EQ1 (kWh)', 'E_charge (kWh)', 'E_discharge (kWh)',
    'SOC_begin (kWh)', 'SOC_end EQ9/12 (kWh)', 'E_grid EQ2-4 (kWh)', 'DeltaE (kWh)',
    'Elasticity EQ6', 'U_cons EQ7 (€)', 'E_actual EQ22 (kWh)', 'D_shift (€)',
    'E_bat EQ11 (kWh)', 'C_deg EQ10 (€)', 'E_sell (kWh)', 'E_buy (kWh)',
    'W_spot (€)', 'Utility EQ8/29 (€)'
]].head(25))
```

```python
# AJ: Roll24 Umax EQ8/29 (€) = SUM(AI2:AI25) - Rolling 24-hour sum for 'Utility EQ8/29 (€)'
# This is a rolling sum over 24 hours, similar to how rolling min/max was calculated.
df_excel['Roll24 Umax EQ8/29 (€)'] = df_excel['Utility EQ8/29 (€)'].rolling(window=24, min_periods=1).sum()

# AK: SDR = IF(AG2=0,IF(AF2>0,1,0),MIN(1,AF2/AG2))
# AG2 is 'E_buy (kWh)', AF2 is 'E_sell (kWh)'
df_excel['SDR'] = np.where(df_excel['E_buy (kWh)'] == 0,
                           np.where(df_excel['E_sell (kWh)'] > 0, 1, 0),
                           np.minimum(1, df_excel['E_sell (kWh)'] / df_excel['E_buy (kWh)']))

# AL: Price_prosumer (€/kWh) = IFERROR(INDEX(L2:L25,MATCH(Q2,H2:H25,0)),N2)
# L2:L25 is rolling Price_spot, Q2 is Roll24 PV Max, H2:H25 is rolling Epv_PV, N2 is Roll24 Price Max
# This needs to find the spot price at the time of maximum PV generation within the rolling 24-hour window.
# Given the complexity and potential for multiple matches or no match, let's simplify based on the intent.
# The Excel formula IFERROR(INDEX(L2:L25,MATCH(Q2,H2:H25,0)),N2) implies finding the Price_spot (L) corresponding
# to the Roll24 PV Max (Q) within the past 24 hours (H) and if not found, use Roll24 Price Max (N).

def calculate_price_prosumer(row):
    window_start_idx = max(0, row.name - 23)
    window = df_excel.loc[window_start_idx : row.name]

    # Find the index of the max PV within the window
    max_pv_value = row['Roll24 PV Max']

    # Check if max_pv_value is 0 or if the PV values in the window are all 0
    if max_pv_value == 0 or (window['Epv _PV,Generation (KWh)'] == 0).all():
        return row['Roll24 Price Max']

    # Find the first occurrence of max_pv_value in the window's 'Epv _PV,Generation (KWh)'
    # Use idxmax for finding the first index of the maximum value
    matching_pv_index = (window['Epv _PV,Generation (KWh)'] == max_pv_value).idxmax()

    # If a match is found, return the corresponding 'Price_spot (€/kWh)'
    if matching_pv_index is not None:
        return df_excel.loc[matching_pv_index, 'Price_spot (€/kWh)']
    else:
        return row['Roll24 Price Max'] # Fallback if no specific match within window

df_excel['Price_prosumer (€/kWh)'] = df_excel.apply(calculate_price_prosumer, axis=1)


# AM: Price_P2P EQ13 (€/kWh) = IFERROR(L2*AL2/(L2-AL2)*Parameters!$C$14+AL2,(L2+AL2)/2)
# L2 is 'Price_spot (€/kWh)', AL2 is 'Price_prosumer (€/kWh)', Parameters!$C$14 is 'SDR'
# The IFERROR handles division by zero or other errors. The Excel formula uses (L2-AL2) as a divisor.
# If L2 == AL2, this would be a division by zero.
price_spot = df_excel['Price_spot (€/kWh)']
price_prosumer = df_excel['Price_prosumer (€/kWh)']
sdr = parameters['SDR']

def calculate_price_p2p(row):
    if row['Price_spot (€/kWh)'] == row['Price_prosumer (€/kWh)']:
        return (row['Price_spot (€/kWh)'] + row['Price_prosumer (€/kWh)']) / 2
    else:
        return row['Price_spot (€/kWh)'] * row['Price_prosumer (€/kWh)'] / \
               (row['Price_spot (€/kWh)'] - row['Price_prosumer (€/kWh)']) * sdr + row['Price_prosumer (€/kWh)']

df_excel['Price_P2P EQ13 (€/kWh)'] = df_excel.apply(calculate_price_p2p, axis=1)

# AN: E_inj EQ16 (kWh) = MAX(0,H2-S2-T2) --> This is the same as AF ('E_sell (kWh)')
# From Excel sheet: "E_inj EQ16 (kWh)" is AN, and AF is "E_sell (kWh)". The formula is the same.
# So, we can just assign the column directly.
df_excel['E_inj EQ16 (kWh)'] = df_excel['E_sell (kWh)']

# AO: U_prosumer EQ14 (€) = Z2-AC2+AN2*AM2-AE2
# Z2 is 'U_cons EQ7 (€)', AC2 is 'D_shift (€)', AN2 is 'E_inj EQ16 (kWh)',
# AM2 is 'Price_P2P EQ13 (€/kWh)', AE2 is 'C_deg EQ10 (€)'
df_excel['U_prosumer EQ14 (€)'] = df_excel['U_cons EQ7 (€)'] - \
                                  df_excel['D_shift (€)'] + \
                                  df_excel['E_inj EQ16 (kWh)'] * df_excel['Price_P2P EQ13 (€/kWh)'] - \
                                  df_excel['C_deg EQ10 (€)']

# AP: a1 EQ18 = Parameters!$C$20+(Parameters!$C$21-Parameters!$C$20)*IF(MAX(I2:I25)=MIN(I2:I25),0,(I2-MIN(I2:I25))/(MAX(I2:I25)-MIN(I2:I25)))
# Parameters!$C$20 is 'a1_min', Parameters!$C$21 is 'a1_max'
# I2 is 'E,base,t Base load Demand_kWh'
# This represents a scaling factor for network cost based on the current load relative to min/max load in a rolling 24h window.

rolling_min_demand = df_excel['E,base,t Base load Demand_kWh'].rolling(window=24, min_periods=1).min()
rolling_max_demand = df_excel['E,base,t Base load Demand_kWh'].rolling(window=24, min_periods=1).max()

df_excel['a1 EQ18'] = parameters['a1_min'] + (parameters['a1_max'] - parameters['a1_min']) * \
                      np.where(rolling_max_demand == rolling_min_demand, 0,
                               (df_excel['E,base,t Base load Demand_kWh'] - rolling_min_demand) / \
                               (rolling_max_demand - rolling_min_demand))

# AQ: Ltotal/Cgrid = IF(MAX(I2:I25)=0,0,I2/MAX(I2:I25))
# I2 is 'E,base,t Base load Demand_kWh'
rolling_max_base_load = df_excel['E,base,t Base load Demand_kWh'].rolling(window=24, min_periods=1).max()
df_excel['Ltotal/Cgrid'] = np.where(rolling_max_base_load == 0, 0,
                                   df_excel['E,base,t Base load Demand_kWh'] / rolling_max_base_load)

# AR: gamma EQ18 = Parameters!$C$15*AP2
# Parameters!$C$15 is 'rho', AP2 is 'a1 EQ18'
df_excel['gamma EQ18'] = parameters['rho'] * df_excel['a1 EQ18']

# AS: tau EQ17 (€/kWh) = Parameters!$C$15+AR2*AQ2^2
# Parameters!$C$15 is 'rho', AR2 is 'gamma EQ18', AQ2 is 'Ltotal/Cgrid'
df_excel['tau EQ17 (€/kWh)'] = parameters['rho'] + df_excel['gamma EQ18'] * df_excel['Ltotal/Cgrid']**2

# AT: Green benefit (€) = Parameters!$C$16*Parameters!$C$17*AN2
# Parameters!$C$16 is 'theta', Parameters!$C$17 is 'rho', AN2 is 'E_inj EQ16 (kWh)'
df_excel['Green benefit (€)'] = parameters['theta'] * parameters['rho'] * df_excel['E_inj EQ16 (kWh)']

# AU: SW_total EQ15 (€) = Z2-AC2-AE2+AT2-AS2*AN2
# Z2 is 'U_cons EQ7 (€)', AC2 is 'D_shift (€)', AE2 is 'C_deg EQ10 (€)',
# AT2 is 'Green benefit (€)', AS2 is 'tau EQ17 (€/kWh)', AN2 is 'E_inj EQ16 (kWh)'
df_excel['SW_total EQ15 (€)'] = df_excel['U_cons EQ7 (€)'] - \
                               df_excel['D_shift (€)'] - \
                               df_excel['C_deg EQ10 (€)'] + \
                               df_excel['Green benefit (€)'] - \
                               df_excel['tau EQ17 (€/kWh)'] * df_excel['E_inj EQ16 (kWh)']

# AV: U_buyer EQ19 (€) = K2*AN2-Parameters!$C$3/2*AN2^2-AM2*AN2+Parameters!$C$16*Parameters!$C$17*AN2
# K2 is 'ω_t (€/kWh)', AN2 is 'E_inj EQ16 (kWh)', Parameters!$C$3 is 'alpha',
# AM2 is 'Price_P2P EQ13 (€/kWh)', Parameters!$C$16 is 'theta', Parameters!$C$17 is 'rho'
df_excel['U_buyer EQ19 (€)'] = df_excel['ω_t (€/kWh)'] * df_excel['E_inj EQ16 (kWh)'] - \
                               parameters['alpha'] / 2 * df_excel['E_inj EQ16 (kWh)']**2 - \
                               df_excel['Price_P2P EQ13 (€/kWh)'] * df_excel['E_inj EQ16 (kWh)'] + \
                               parameters['theta'] * parameters['rho'] * df_excel['E_inj EQ16 (kWh)']

# AW: A EQ21 (€) = K2*AN2-Parameters!$C$3/2*AN2^2+Parameters!$C$16*Parameters!$C$17*AN2-K2*AA2+Parameters!$C$3/2*AA2^2+AC2+AE2
# K2 is 'ω_t (€/kWh)', AN2 is 'E_inj EQ16 (kWh)', Parameters!$C$3 is 'alpha',
# Parameters!$C$16 is 'theta', Parameters!$C$17 is 'rho',
# AA2 is 'E_actual EQ22 (kWh)', AC2 is 'D_shift (€)', AE2 is 'C_deg EQ10 (€)'
df_excel['A EQ21 (€)'] = df_excel['ω_t (€/kWh)'] * df_excel['E_inj EQ16 (kWh)'] - \
                         parameters['alpha'] / 2 * df_excel['E_inj EQ16 (kWh)']**2 + \
                         parameters['theta'] * parameters['rho'] * df_excel['E_inj EQ16 (kWh)'] - \
                         df_excel['ω_t (€/kWh)'] * df_excel['E_actual EQ22 (kWh)'] + \
                         parameters['alpha'] / 2 * df_excel['E_actual EQ22 (kWh)']**2 + \
                         df_excel['D_shift (€)'] + \
                         df_excel['C_deg EQ10 (€)']

# AX: Price_eq EQ20 (€/kWh) = IF(AN2=0,"",AW2/(2*AN2))
# AN2 is 'E_inj EQ16 (kWh)', AW2 is 'A EQ21 (€)'
df_excel['Price_eq EQ20 (€/kWh)'] = np.where(df_excel['E_inj EQ16 (kWh)'] == 0, np.nan,
                                            df_excel['A EQ21 (€)'] / (2 * df_excel['E_inj EQ16 (kWh)']))

# AY: CS EQ23 (€) = (K2+Parameters!$C$16*Parameters!$C$17-AM2)*AN2-Parameters!$C$3/2*AN2^2
# K2 is 'ω_t (€/kWh)', Parameters!$C$16 is 'theta', Parameters!$C$17 is 'rho',
# AM2 is 'Price_P2P EQ13 (€/kWh)', AN2 is 'E_inj EQ16 (kWh)', Parameters!$C$3 is 'alpha'
df_excel['CS EQ23 (€)'] = (df_excel['ω_t (€/kWh)'] + parameters['theta'] * parameters['rho'] - \
                           df_excel['Price_P2P EQ13 (€/kWh)']) * df_excel['E_inj EQ16 (kWh)'] - \
                           parameters['alpha'] / 2 * df_excel['E_inj EQ16 (kWh)']**2

# AZ: PS EQ24 (€) = AN2*AM2-Parameters!$C$18*AN2-AE2-Parameters!$C$4*ABS(AB2)*AN2-Parameters!$C$19/2*AN2^2
# AN2 is 'E_inj EQ16 (kWh)', AM2 is 'Price_P2P EQ13 (€/kWh)', Parameters!$C$18 is 'b_i',
# AE2 is 'C_deg EQ10 (€)', Parameters!$C$4 is 'zeta', AB2 (delta_e_ab_calc) is 'E,base,t Base load Demand_kWh' - 'E_actual EQ22 (kWh)' (from previous step)
# Parameters!$C$19 is 'a_i'
df_excel['PS EQ24 (€)'] = df_excel['E_inj EQ16 (kWh)'] * df_excel['Price_P2P EQ13 (€/kWh)'] - \
                           parameters['b_i'] * df_excel['E_inj EQ16 (kWh)'] - \
                           df_excel['C_deg EQ10 (€)'] - \
                           parameters['zeta'] * abs(delta_e_ab_calc) * df_excel['E_inj EQ16 (kWh)'] - \
                           parameters['a_i'] / 2 * df_excel['E_inj EQ16 (kWh)']**2

# BA: LC EQ26 (€) = (K2+Parameters!$C$16*Parameters!$C$17-Parameters!$C$18-Parameters!$C$5-Parameters!$C$4*ABS(AB2))*AN2
# K2 is 'ω_t (€/kWh)', Parameters!$C$16 is 'theta', Parameters!$C$17 is 'rho',
# Parameters!$C$18 is 'b_i', Parameters!$C$5 is 'psi', Parameters!$C$4 is 'zeta',
# AB2 (delta_e_ab_calc) is 'E,base,t Base load Demand_kWh' - 'E_actual EQ22 (kWh)', AN2 is 'E_inj EQ16 (kWh)'
df_excel['LC EQ26 (€)'] = (df_excel['ω_t (€/kWh)'] + parameters['theta'] * parameters['rho'] - \
                           parameters['b_i'] - parameters['psi'] - \
                           parameters['zeta'] * abs(delta_e_ab_calc)) * df_excel['E_inj EQ16 (kWh)']

# BB: QC EQ27 (€) = (Parameters!$C$3+Parameters!$C$19)/2*AN2^2
# Parameters!$C$3 is 'alpha', Parameters!$C$19 is 'a_i', AN2 is 'E_inj EQ16 (kWh)'
df_excel['QC EQ27 (€)'] = (parameters['alpha'] + parameters['a_i']) / 2 * df_excel['E_inj EQ16 (kWh)']**2

# BC: SW_P2P EQ25 (€) = BA2-BB2
# BA2 is 'LC EQ26 (€)', BB2 is 'QC EQ27 (€)'
df_excel['SW_P2P EQ25 (€)'] = df_excel['LC EQ26 (€)'] - df_excel['QC EQ27 (€)']

# BD: CS_pros EQ28 (€) = IF(R2="CHARGE",X2*(K2+Parameters!$C$3*Parameters!$C$7-Parameters!$C$3*H2-L2)-Parameters!$C$3/2*X2^2,
#                          IF(R2="DISCHARGE",X2*(K2-Parameters!$C$3*Parameters!$C$8-Parameters!$C$3*H2-L2)-Parameters!$C$3/2*X2^2,
#                          X2*(K2-Parameters!$C$3*H2-L2)-Parameters!$C$3/2*X2^2))
# R2 is 'Decision', X2 is 'DeltaE (kWh)', K2 is 'ω_t (€/kWh)', Parameters!$C$3 is 'alpha',
# Parameters!$C$7 is 'P_ch_max', Parameters!$C$8 is 'P_dis_max',
# H2 is 'Epv _PV,Generation (KWh)', L2 is 'Price_spot (€/kWh)'

charge_condition = (df_excel['Decision'] == 'CHARGE')
discharge_condition = (df_excel['Decision'] == 'DISCHARGE')

# Base calculation for the ELSE case
base_cs_pros = df_excel['DeltaE (kWh)'] * (df_excel['ω_t (€/kWh)'] - parameters['alpha'] * df_excel['Epv _PV,Generation (KWh)'] - df_excel['Price_spot (€/kWh)']) - \
               parameters['alpha'] / 2 * df_excel['DeltaE (kWh)']**2

# CHARGE case
charge_cs_pros = df_excel['DeltaE (kWh)'] * (df_excel['ω_t (€/kWh)'] + parameters['alpha'] * parameters['P_ch_max'] - parameters['alpha'] * df_excel['Epv _PV,Generation (KWh)'] - df_excel['Price_spot (€/kWh)']) - \
                 parameters['alpha'] / 2 * df_excel['DeltaE (kWh)']**2

# DISCHARGE case
discharge_cs_pros = df_excel['DeltaE (kWh)'] * (df_excel['ω_t (€/kWh)'] - parameters['alpha'] * parameters['P_dis_max'] - parameters['alpha'] * df_excel['Epv _PV,Generation (KWh)'] - df_excel['Price_spot (€/kWh)']) - \
                    parameters['alpha'] / 2 * df_excel['DeltaE (kWh)']**2

df_excel['CS_pros EQ28 (€)'] = np.where(charge_condition, charge_cs_pros,
                                       np.where(discharge_condition, discharge_cs_pros,
                                                base_cs_pros))

# BE: CSt1 EQ30 (€) = X2*(K2-Parameters!$C$3*Parameters!$C$8-Parameters!$C$3*H2-L2)-Parameters!$C$3/2*X2^2
# X2 is 'DeltaE (kWh)', K2 is 'ω_t (€/kWh)', Parameters!$C$3 is 'alpha',
# Parameters!$C$8 is 'P_dis_max', H2 is 'Epv _PV,Generation (KWh)', L2 is 'Price_spot (€/kWh)'
df_excel['CSt1 EQ30 (€)'] = df_excel['DeltaE (kWh)'] * (df_excel['ω_t (€/kWh)'] - parameters['alpha'] * parameters['P_dis_max'] - parameters['alpha'] * df_excel['Epv _PV,Generation (KWh)'] - df_excel['Price_spot (€/kWh)']) - \
                            parameters['alpha'] / 2 * df_excel['DeltaE (kWh)']**2

# BF: Price_peak24 (€/kWh) = MAX(L2:L25) --> This is the same as N ('Roll24 Price Max')
df_excel['Price_peak24 (€/kWh)'] = df_excel['Roll24 Price Max']

# BG: GS_grid EQ31 (€) = MAX(0,(BF2-L2)*U2)
# BF2 is 'Price_peak24 (€/kWh)', L2 is 'Price_spot (€/kWh)', U2 is 'E_discharge (kWh)'
df_excel['GS_grid EQ31 (€)'] = np.maximum(0,
                                         (df_excel['Price_peak24 (€/kWh)'] - df_excel['Price_spot (€/kWh)']) * \
                                         df_excel['E_discharge (kWh)'])

# BH: SW_grid EQ32 (€) = BD2+BG2
# BD2 is 'CS_pros EQ28 (€)', BG2 is 'GS_grid EQ31 (€)'
df_excel['SW_grid EQ32 (€)'] = df_excel['CS_pros EQ28 (€)'] + df_excel['GS_grid EQ31 (€)']

print("Οι στήλες AJ έως BH υπολογίστηκαν επιτυχώς.")
print(df_excel.head())
```

```python
# Implement Daily and Annual Aggregations

# Create a 'Date' column from the 'Timestamp' for daily aggregation
df_excel['Date'] = pd.to_datetime(df_excel['Timestamp']).dt.date

# Daily Aggregations
daily_summary = df_excel.groupby('Date').agg(
    Daily_Umax_EQ8_29=('Utility EQ8/29 (€)', 'sum'),
    Daily_Uprosumer_EQ14=('U_prosumer EQ14 (€)', 'sum'),
    Daily_SWtotal_EQ15=('SW_total EQ15 (€)', 'sum'),
    Daily_CS_EQ23=('CS EQ23 (€)', 'sum'),
    Daily_PS_EQ24=('PS EQ24 (€)', 'sum'),
    Daily_SWP2P_EQ25=('SW_P2P EQ25 (€)', 'sum'),
    Daily_CSpros_EQ28=('CS_pros EQ28 (€)', 'sum'),
    Daily_SWgrid_EQ32=('SW_grid EQ32 (€)', 'sum')
).reset_index()

# Annual Aggregations
annual_summary = pd.DataFrame({
    'Annual KPI': [
        'Annual Umax EQ8/29',
        'Annual Uprosumer EQ14',
        'Annual SWtotal EQ15',
        'Annual CS EQ23',
        'Annual PS EQ24',
        'Annual SWP2P EQ25',
        'Annual CSpros EQ28',
        'Annual SWgrid EQ32'
    ],
    'Value (€)': [
        daily_summary['Daily_Umax_EQ8_29'].sum(),
        daily_summary['Daily_Uprosumer_EQ14'].sum(),
        daily_summary['Daily_SWtotal_EQ15'].sum(),
        daily_summary['Daily_CS_EQ23'].sum(),
        daily_summary['Daily_PS_EQ24'].sum(),
        daily_summary['Daily_SWP2P_EQ25'].sum(),
        daily_summary['Daily_CSpros_EQ28'].sum(),
        daily_summary['Daily_SWgrid_EQ32'].sum()
    ]
})

print("Ημερήσιες Συγκεντρωτικές Αναφορές:")
display(daily_summary.head())
print("Ετήσιες Συγκεντρωτικές Αναφορές:")
display(annual_summary)
```
